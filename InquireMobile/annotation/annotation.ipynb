```python
def convert_file_to_utf8(input_file_path, output_file_path):
    with open(input_file_path, 'r', encoding='iso-8859-1') as infile:
        data = infile.read()

    with open(output_file_path, 'w', encoding='utf-8') as outfile:
        outfile.write(data)

input_file_path = "/home/aiqihang.aqh/Appagent/【iTAG】不皮_人工介入YN判断3.0_GBK__20250523114232.jsonl"
output_file_path = "/home/aiqihang.aqh/Appagent/output_utf8.jsonl"

convert_file_to_utf8(input_file_path, output_file_path)
```
