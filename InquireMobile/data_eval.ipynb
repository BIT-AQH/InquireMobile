```python
import pandas as pd
from tqdm import tqdm
import json,re
import os
import matplotlib.pyplot as plt
import seaborn as sns
import shutil
from PIL import Image
import math

def read_jsonl(file_path):
    data = []
    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            try:
                json_object = json.loads(line.strip())
                data.append(json_object)
            except json.JSONDecodeError as e:
                print(f"Error decoding JSON: {e}")
                continue
                # 如果选择抛出异常，使用下面这行
                # raise
    return data
```

```python
# split en eval data: huawei and simulator

input_data_path = "/home/aiqihang.aqh/Appagent/data/tasks_en.jsonl"
en_data = read_jsonl(input_data_path)

virtual_apps = ["x","YouTube","microsoft word","bilibili","google map", "taobao","phone","contacts","pinduoduo","jd","youtube","instagram","settings","video player","music player"]

virtual_test_data = []
huawei_test_data = []

new_en_data = []

for data in en_data:

    is_simulator_available = False

    for app in data['apps'] + data['intension'].split(" "):
        if app in virtual_apps:
            is_simulator_available = True


    if is_simulator_available:
        virtual_test_data.append(data)
    else:

        huawei_test_data.append(data)


print(len(virtual_test_data))
    
print(len(huawei_test_data))

with open("/home/aiqihang.aqh/Appagent/data/tasks_en_huawei.jsonl", "w", encoding='utf-8') as file:
    for data in huawei_test_data:
        json.dump(data, file, ensure_ascii=False)
        file.write("\n")

with open("/home/aiqihang.aqh/Appagent/data/tasks_en_simulator.jsonl", "w", encoding='utf-8') as file:
    for data in virtual_test_data:
        json.dump(data, file, ensure_ascii=False)
        file.write("\n")
```
